## Project overview
This notebook demonstrates a reproducible sales analytics workflow using a clearly labeled synthetic dataset. It explores sales trends, product and regional performance, and simple business insights. The generated data is for learning and demonstration only; it is not real company data.

## Objectives
1. Create and inspect a sample sales dataset.
2. Clean and prepare data for analysis.
3. Calculate revenue and summarize performance by month, region, and product category.
4. Visualize key patterns and communicate actionable findings.

## Dataset description
The notebook generates 240 sample transactions with order dates, regions, product categories, units sold, and unit prices. Revenue is calculated as units sold multiplied by unit price.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
np.random.seed(42)

# Generate a reproducible synthetic dataset (not real business data).
n = 240
dates = pd.date_range('2025-01-01', '2025-12-31', periods=n)
regions = np.random.choice(['North', 'South', 'East', 'West'], size=n)
categories = np.random.choice(['Electronics', 'Home & Kitchen', 'Clothing', 'Books'], size=n,
                               p=[0.30, 0.25, 0.25, 0.20])
price_ranges = {'Electronics': (800, 5000), 'Home & Kitchen': (250, 2500),
                'Clothing': (300, 1800), 'Books': (100, 800)}
units = np.random.randint(1, 9, size=n)
unit_prices = [round(np.random.uniform(*price_ranges[c]), 2) for c in categories]

df = pd.DataFrame({
    'OrderDate': dates,
    'Region': regions,
    'Category': categories,
    'UnitsSold': units,
    'UnitPrice': unit_prices
})
df['Revenue'] = (df['UnitsSold'] * df['UnitPrice']).round(2)
df.head()
```

## 1. Inspect and clean the data
Check data types, missing values, and duplicate rows before analysis.

```python
print('Dataset shape:', df.shape)
display(df.info())
print('\nMissing values by column:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())

# Basic preparation: parse dates, remove exact duplicates, and discard incomplete rows.
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce')
df = df.drop_duplicates().dropna(subset=['OrderDate', 'Region', 'Category', 'UnitsSold', 'UnitPrice'])
df['Revenue'] = (df['UnitsSold'] * df['UnitPrice']).round(2)
df['Month'] = df['OrderDate'].dt.to_period('M').astype(str)
print('Cleaned dataset shape:', df.shape)
df.describe(include='all')
```

## 2. Key performance indicators
These metrics summarize the overall sample dataset.

```python
total_revenue = df['Revenue'].sum()
total_units = df['UnitsSold'].sum()
average_order_revenue = df['Revenue'].mean()
transaction_count = len(df)

print(f'Total revenue: ₹{total_revenue:,.2f}')
print(f'Total units sold: {total_units:,}')
print(f'Average transaction revenue: ₹{average_order_revenue:,.2f}')
print(f'Number of transactions: {transaction_count:,}')
```

## 3. Monthly revenue trend
A line chart helps reveal changes in revenue over the year.

```python
monthly_revenue = df.groupby('Month', as_index=False)['Revenue'].sum()
plt.figure(figsize=(12, 5))
plt.plot(monthly_revenue['Month'], monthly_revenue['Revenue'], marker='o')
plt.title('Monthly Revenue (Synthetic Sample Data)')
plt.xlabel('Month')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
monthly_revenue
```

## 4. Revenue by product category
Compare category-level revenue to understand which categories contribute most.

```python
category_revenue = df.groupby('Category', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
plt.figure(figsize=(9, 5))
plt.bar(category_revenue['Category'], category_revenue['Revenue'])
plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
category_revenue
```

## 5. Regional performance
Compare total revenue across regions.

```python
region_revenue = df.groupby('Region', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
plt.figure(figsize=(8, 5))
plt.bar(region_revenue['Region'], region_revenue['Revenue'])
plt.title('Revenue by Region')
plt.xlabel('Region')
plt.ylabel('Revenue (₹)')
plt.tight_layout()
plt.show()
region_revenue
```

## 6. Business insights
The following statements are generated from the sample data when the notebook runs. They should not be treated as findings about a real company.

```python
best_category = category_revenue.iloc[0]
best_region = region_revenue.iloc[0]
best_month = monthly_revenue.loc[monthly_revenue['Revenue'].idxmax()]
print(f"Highest-revenue category in this sample: {best_category['Category']} (₹{best_category['Revenue']:,.2f})")
print(f"Highest-revenue region in this sample: {best_region['Region']} (₹{best_region['Revenue']:,.2f})")
print(f"Highest-revenue month in this sample: {best_month['Month']} (₹{best_month['Revenue']:,.2f})")

print('\nSuggested actions for a real business after validating with actual data:')
print('- Investigate drivers of the highest-performing category and region.')
print('- Compare monthly patterns with promotions, seasonality, and stock availability.')
print('- Review low-performing categories before changing inventory or marketing spend.')
```

## 7. Export summary tables
Save the key summary tables as CSV files in the current working directory.

```python
monthly_revenue.to_csv('monthly_revenue_summary.csv', index=False)
category_revenue.to_csv('category_revenue_summary.csv', index=False)
region_revenue.to_csv('region_revenue_summary.csv', index=False)
print('Exported monthly_revenue_summary.csv, category_revenue_summary.csv, and region_revenue_summary.csv')
```
